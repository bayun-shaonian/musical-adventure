```python
# 程序分析
# 1：
import random
import math
n = random.randint(1, 100)  # 函数 random.randint(a,b) 返回从 a 到 b 之间随机选出的一个整数
a, b, cnt = 1, 100, 0
while cnt < n:
    cnt += 1
    m = math.ceil((a + b) / 2)  # 函数 math.ceil(x) 的返回值为数值 x 的向上取整
    if n < m:
        b = m - 1
    elif n>m:
        a = m
    else :
        cnt += 1
print(cnt)
```

```python
# 3:
def Change(radix, num):
    g = []
    while num != 0:
        r = num % radix
        num =num // radix
        g.append(r)
    ans = ''
    t = ('A', 'B', 'C', 'D', 'E', 'F')
    for x in range(len(g)):
        if g[x] >= 10:
            ans +=t[g[x]-10]
        else:
            ans += str(g[x])
    return ans[::-1]
print(Change(16, 12455))
# 程序输出：30A7
```

```python
# 4:
def identical(L):
    flag = 0
    L.sort()
    for i in range(len(L)-1):
        if L[i] != L[i+1]:
            flag = 1
            break
    if flag:
        return True
    else:
        return False
```

```python
# 数据分析：
import numpy as py
import pandas as pd
import glob
driver=pd.read_csv('driver.csv')
average_driver_rating = driver['rating'].mean()
task1 = {'insight_type': 'average_driver_rating', 'value': round(average_driver_rating, 2)}
```

```python
has_second_language = driver[driver['second_language'] != 'no'].shape[0]
total_drivers = driver.shape[0]
percentage_drivers_with_second_language = (has_second_language / total_drivers) * 100

task2 = {'insight_type': 'percentage_drivers_with_second_language', 'value': round(percentage_drivers_with_second_language, 2)}
```

```python
files = glob.glob('rides_*.csv')
df_list = []
for file in files:
    df = pd.read_csv(file)
    df_list.append(df)
merged_df = pd.concat(df_list, ignore_index=True)
success_rides = merged_df[merged_df['status'] == 'Success'].shape[0]
total_rides = merged_df.shape[0]
ride_success_rate = (success_rides / total_rides) * 100
task3={'insight_type': 'ride_success_rate','value': round(ride_success_rate, 2)}
```

```python
results = pd.DataFrame([task1, task2, task3])
results.to_csv('analysis_results.csv', index=False)
```

```python
# 深度学习训练相关名词解释
1：层与层之间的标准化处理

2：有很好的泛化性

3：梯度裁剪，在反向传播过程中裁剪梯度，使它们永远不会超过某个阈值。这种技术通常用于循环神经网络（RNN），它很难使用批量归一化

4：就是在验证误差达到最小值时停止训练，该方法叫作早停法。通过早停法，一旦验证误差达到最小值就立刻停止训练。这是一个非常简单而有效的正则化技巧

5：防止过拟合
6：反向传播算法的第二阶段从输出层到输入层，沿途传播误差梯度。一旦算法计算出代价函数相对于网络中每个参数的梯度，就可以通过梯度下降步骤使用这些梯度来更新每个参数。

7：如果将学习率设置得太大，训练可能会发散；如果学习率设置得太小，训练会花费很长时间。如果学习率设置得稍大了点，它一开始会很快，但是最终会围绕最优解振荡，不会真正稳定下来
可以对模型进行数百次迭代训练，然后将学习率从很小的值成指数形式地增加到很大的值
```

```python
# 假设我们用 60 天的价格预测第 61 天价格
import numpy as np
time_steps = 60
features = 1
samples = 200

# 构造输入数据 (samples, time_steps, features)
x_train = np.random.rand(samples, time_steps, features)  # 200个 60天股票价格
y_train = np.random.rand(samples, 1)                     # 200个 对应的第61天股票价格
x_test  = np.random.rand(10, time_steps, features)       # 测试
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dropout, Dense
time_steps = 60
features = 1
stock_model = Sequential()
stock_model.add(LSTM(units=4, input_shape=(time_steps, features), return_sequences=True))
stock_model.add(Dropout(rate=0.2))
stock_model.add(LSTM(units=4, return_sequences=False))
stock_model.add(Dropout(rate=0.2))
stock_model.add(Dense(units=2))
stock_model.add(Dense(units=1))
stock_model.compile(loss='mean_squared_error', optimizer='adam')
stock_model.fit(x_train, y_train, epochs=5, batch_size=16, verbose=0)
stock_model.summary()
predictions = stock_model.predict(x_test)
```
